```python
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
```

```python
import numpy as np
x = np.linspace(0,5,11)
y = x ** 2
x
```

```python
y
```

```python
# Functional
plt.plot(x,y)
```

```python
plt.plot(x, y)
plt.xlabel("X Label")
plt.ylabel("Y Label")
plt.title("Title")
```

```python
plt.subplot(1,2,1)
plt.plot(x,y,"r")
plt.subplot(1,2,2)
plt.plot(y,x,"b")
```

```python
# OOP
fig = plt.figure()
axes = fig.add_axes((0.1,0.1,0.8,0.8))
axes.plot(x,y)
axes.set_xlabel("X Label")
axes.set_ylabel("Y Label")
axes.set_title("Set Title")

```

```python
fig = plt.figure()
axes1 = fig.add_axes((0.1,0.1,0.8,0.8))
axes2 = fig.add_axes((0.2,0.5,0.4,0.4))
axes1.plot(x,y)
axes1.set_title("Larger Plot")
axes1.set_xlabel("X Label")
axes1.set_ylabel("Y Label")

axes2.plot(y,x)
axes2.set_title("Smaller Plot")
axes2.set_ylabel("Y Label")
axes2.set_xlabel("X Label")
```

```python
fig = plt.figure()
axes1 = fig.add_axes((0.1,0.1,0.8,0.8))
axes1.plot(x,y)
```

```python
fig,axes = plt.subplots()
axes.plot(x,y)
```

```python
fig, axes = plt.subplots(nrows=3, ncols=3)
for current_ax in axes.flat:
    current_ax.plot(x, y)
plt.tight_layout()
```

```python
fig, axes = plt.subplots(nrows=1, ncols=2)
axes[0].plot(x,y)
axes[0].set_title("Test")
```

```python
fig = plt.figure(figsize=(3,2),dpi=100)
ax = fig.add_axes((0,0,1,1))
ax.plot(x,y)
```

```python
fig, axes = plt.subplots(figsize=(8,2))
axes.plot(x,y)
```

```python
fig = plt.figure()
ax = fig.add_axes((0,0,1,1))
ax.plot(x,y)
ax.plot(y,x)
```

```python
fig = plt.figure()
ax = fig.add_axes((0, 0, 1, 1))
ax.plot(x, y**2)
ax.plot(y, x**3)
```

```python
fig = plt.figure()
ax = fig.add_axes((0, 0, 1, 1))
ax.plot(x, y**2,label="X Squared")
ax.plot(y, x**3, label="X Cubed")
ax.legend(loc=0)
```
