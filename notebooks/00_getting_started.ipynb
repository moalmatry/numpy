# Getting Started with NumPy

Starter notebook to verify your `uv` + Jupyter environment and inspect basic `ndarray` properties.

```python
import numpy as np
from numpy_study import describe

print(f"NumPy version: {np.__version__}")
```

```python
arr = np.arange(1, 13).reshape(3, 4)
describe(arr)
```

```python
print("Row sums:", arr.sum(axis=1))
print("Col means:", arr.mean(axis=0))
print("Squared:\n", arr ** 2)
```
