```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 5, 11)
y = x**2
```

```python
fig = plt.figure()
ax = fig.add_axes((0, 0, 1, 1))
ax.plot(x, y, color="#aaafff", linewidth=3, alpha=0.5, linestyle="-.")
```

```python
fig = plt.figure()
ax = fig.add_axes((0, 0, 1, 1))
ax.plot(
    x,
    y,
    color="#aaafff",
    linewidth=3,
    marker="o",
    markersize="20",
    markerfacecolor="yellow",
    markeredgewidth=5,
    markeredgecolor="red",
)
```

```python
fig = plt.figure()
ax = fig.add_axes((0, 0, 1, 1))
ax.plot(x, y, color="purple", lw=3, ls="--")
ax.set_xlim((0, 1))
ax.set_ylim((0, 2))
```
